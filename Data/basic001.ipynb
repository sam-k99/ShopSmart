```python
import pandas as pd

# Load Data
df = pd.read_csv('/home/sam/projects/ShopSmart/Data/online_retail_II.csv')

# Fix Types
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Customer ID'] = df['Customer ID'].fillna('GUEST').astype(str)

# Filter Valid Transactions (Remove Cancellations)
valid_df = df[~df['Invoice'].str.startswith('C', na=False)].copy()

# Calculate Revenue
valid_df['Revenue'] = valid_df['Quantity'] * valid_df['Price']

print(f"✅ Ready: {len(valid_df):,} valid transactions loaded")
```

```python
import pandas as pd

# Load Data
df = pd.read_csv('/home/sam/projects/ShopSmart/Data/online_retail_II.csv')

# Fix Types
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Customer ID'] = df['Customer ID'].fillna('GUEST').astype(str)

# Filter Valid Transactions (Remove Cancellations)
valid_df = df[~df['Invoice'].str.startswith('C', na=False)].copy()

# Calculate Revenue
valid_df['Revenue'] = valid_df['Quantity'] * valid_df['Price']

print(f"✅ Ready: {len(valid_df):,} valid transactions loaded")
```

```python
# Tag Sales vs Adjustments
valid_df['Transaction_Type'] = 'Sale'
valid_df.loc[valid_df['Revenue'] < 0, 'Transaction_Type'] = 'Adjustment'

# Verify segmentation
print("=== Transaction Type Breakdown ===")
print(valid_df['Transaction_Type'].value_counts())
print(f"\nCore Sales Revenue: ${valid_df[valid_df['Transaction_Type']=='Sale']['Revenue'].sum():,.2f}")
print(f"Total Adjustments: ${valid_df[valid_df['Transaction_Type']=='Adjustment']['Revenue'].sum():,.2f}")
```

```python
# Monthly Revenue Trend (Core Sales Only)
monthly_sales = valid_df[valid_df['Transaction_Type']=='Sale'].groupby(
    pd.Grouper(key='InvoiceDate', freq='ME')
)['Revenue'].sum()

print("=== Monthly Core Sales Revenue ===")
print(monthly_sales.head(12))  # First year
print(f"\nAvg Monthly Revenue: ${monthly_sales.mean():,.2f}")
print(f"Peak Month: {monthly_sales.idxmax().strftime('%B %Y')} (${monthly_sales.max():,.2f})")
```

```python
# Define Historical Baseline (Summer 2010: Jun-Aug)
baseline_2010 = monthly_sales.loc['2010-06':'2010-08'].mean()

# Define Campaign Period (Summer 2011: Jun-Aug)
campaign_2011 = monthly_sales.loc['2011-06':'2011-08']
campaign_avg = campaign_2011.mean()

# Calculate Lift
lift_pct = ((campaign_avg - baseline_2010) / baseline_2010) * 100

print("=== SUMMER GLOW CAMPAIGN EVALUATION ===")
print(f"Historical Summer Baseline (2010): ${baseline_2010:,.2f}/mo")
print(f"Campaign Performance (2011):       ${campaign_avg:,.2f}/mo")
print(f"Revenue Lift:                      {lift_pct:+.1f}%")
print(f"\nVerdict: {'✅ SUCCESS' if lift_pct > 5 else '⚠️ FLAT/DECLINE'}")
```

```python
# Filter to campaign period only
campaign_df = valid_df[
    (valid_df['Transaction_Type']=='Sale') & 
    (valid_df['InvoiceDate'] >= '2011-06-01') & 
    (valid_df['InvoiceDate'] <= '2011-08-31')
]

# Top 10 Products by Revenue During Campaign
top_products = campaign_df.groupby('Description')['Revenue'].sum().nlargest(10)

print("=== TOP 10 PRODUCTS DURING SUMMER GLOW ===")
print(top_products)
print(f"\nTotal Campaign Revenue from Top 10: ${top_products.sum():,.2f}")
print(f"% of Total Campaign Revenue: {(top_products.sum() / campaign_df['Revenue'].sum() * 100):.1f}%")
```

```python
# Remove non-product entries
clean_campaign = campaign_df[
    ~campaign_df['Description'].isin(['DOTCOM POSTAGE', 'POSTAGE', 'Manual'])
]

# Re-rank top 10 REAL products
real_top_10 = clean_campaign.groupby('Description')['Revenue'].sum().nlargest(10)
print("=== TOP 10 REAL PRODUCTS (Excl. Postage/Manual) ===")
print(real_top_10)
new_share = (real_top_10.sum() / clean_campaign['Revenue'].sum()) * 100
print(f"\nRevised Top 10 Share: {new_share:.1f}%")
```

```python
# RFM Analysis (Core Sales Only, Excl. GUEST)
rfm_df = valid_df[
    (valid_df['Transaction_Type']=='Sale') & 
    (valid_df['Customer ID'] != 'GUEST')
].copy()

# Calculate RFM metrics
snapshot_date = rfm_df['InvoiceDate'].max() + pd.Timedelta(days=1)
rfm = rfm_df.groupby('Customer ID').agg(
    Recency=('InvoiceDate', lambda x: (snapshot_date - x.max()).days),
    Frequency=('Invoice', 'nunique'),
    Monetary=('Revenue', 'sum')
).reset_index()

# Score each dimension (1-5, 5=best)
for col in ['Recency', 'Frequency', 'Monetary']:
    rfm[f'{col}_Score'] = pd.qcut(rfm[col], q=5, labels=[5,4,3,2,1] if col=='Recency' else [1,2,3,4,5]).astype(int)

rfm['RFM_Score'] = rfm['Recency_Score'].astype(str) + rfm['Frequency_Score'].astype(str) + rfm['Monetary_Score'].astype(str)

print(f"=== RFM SEGMENTATION COMPLETE ===")
print(f"Total Customers Analyzed: {len(rfm):,}")
print(f"\nTop 5 Champions (555):")
print(rfm[rfm['RFM_Score']=='555'].head())
print(f"\nBottom 5 At-Risk (111):")
print(rfm[rfm['RFM_Score']=='111'].head())
```

```python
# Score each dimension with duplicate handling
for col in ['Recency', 'Frequency', 'Monetary']:
    if col == 'Recency':
        labels = [5,4,3,2,1]  # Lower days = better score
    else:
        labels = [1,2,3,4,5]  # Higher values = better score
    
    rfm[f'{col}_Score'] = pd.qcut(
        rfm[col], 
        q=5, 
        labels=labels, 
        duplicates='drop'  # ← CRITICAL FIX FOR SKEWED DATA
    ).astype(int)

# Create composite RFM score
rfm['RFM_Score'] = (
    rfm['Recency_Score'].astype(str) + 
    rfm['Frequency_Score'].astype(str) + 
    rfm['Monetary_Score'].astype(str)
)

print(f"=== RFM SEGMENTATION COMPLETE ===")
print(f"Total Customers Analyzed: {len(rfm):,}")
print(f"\nTop 5 Champions (555):")
print(rfm[rfm['RFM_Score']=='555'].head())
print(f"\nBottom 5 At-Risk (111):")
print(rfm[rfm['RFM_Score']=='111'].head())
```
