```python
import pandas as pd
import os

# Load data with absolute path
df = pd.read_csv('/home/sam/projects/ShopSmart/Data/online_retail_II.csv')

print(f"✅ Data loaded: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
print(f"\nFirst 5 rows:\n{df.head()}")
```

```python
valid_df = df[~df['Invoice'].str.startswith('C', na=False)].copy()
```

```python
valid_df['Revenue'] = valid_df['Quantity'] * valid_df['Price']
```

```python
print(f"\nValid transactions: {len(valid_df):,} rows")
print(f"Revenue range: ${valid_df['Revenue'].min():,.2f} → ${valid_df['Revenue'].max():,.2f}")
print(f"Negative revenue rows: {(valid_df['Revenue'] < 0).sum():,}")
```
